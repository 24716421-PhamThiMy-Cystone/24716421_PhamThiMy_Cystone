# PHẦN B — DESCRIPTIVE STATISTICS
## Group 1 — Central Tendency

```python
# TODO

display(df[['age', 'fare', 'family_size']].mean())
display(df[['age', 'fare', 'family_size']].median())
display(df[['age', 'fare', 'family_size']].mode())
```

## Group 2 — Dispersion

```python
# TODO

display(df[['age', 'fare', 'family_size']].describe())
for col in ['age', 'fare', 'family_size']:
    s = df[col].dropna()
    range_ = s.max() - s.min()
    std_ = s.std()
    q1, q3 = s.quantile([0.25, 0.75])
    iqr = q3 - q1
    cv = std_ / s.mean()
    display(col, range_, std_, iqr, cv)
```

## Group 3 — Location and Shape

```python
# TODO

for col in ['age', 'fare']:
    s = df[col].dropna()
    p25, p50, p75 = s.quantile([0.25, 0.5, 0.75])
    skew_ = stats.skew(s)
    kurt_ = stats.kurtosis(s)
    display(col, p25, p50, p75, skew_, kurt_)
```

# PHẦN C — DEFINE THE QUESTION

## Câu hỏi 1: Hạng vé nào có tỷ lệ sống sót cao nhất, chênh lệch bao nhiêu so với hạng thấp nhất?

```python
# TODO

survival_by_class = df.groupby('class')['survived'].mean()
diff_class = survival_by_class.max() - survival_by_class.min()
display(survival_by_class.round(4))
print(f"\n=> {survival_by_class.idxmax()} có tỷ lệ sống sót cao nhất ({survival_by_class.max()*100:.1f}%), chênh lệch so với hạng thấp nhất là {diff_class*100:.1f}%.")
```

## Câu hỏi 2: Giới tính hay hạng vé ảnh hưởng đến sống sót mạnh hơn?

```python
# TODO

diff_sex = df.groupby('sex')['survived'].mean().max() - df.groupby('sex')['survived'].mean().min()
display(df.groupby('sex')['survived'].mean().round(4))
print(f"Chênh lệch theo Giới tính: {diff_sex*100:.1f}%")
print(f"Chênh lệch theo Hạng vé: {diff_class*100:.1f}%")
print(f"\n=> Giới tính ảnh hưởng mạnh hơn hạng vé đến khả năng sống sót.")
```

## Câu hỏi 3: Vé đắt hơn có thực sự sống sót cao hơn không?

```python
# TODO

fare_by_survived = df.groupby('survived')['fare'].mean()
display(fare_by_survived.round(2))
print(f"\n=> giá vé trung bình của người sống sót ({fare_by_survived[1]:.2f}) cao gấp {(fare_by_survived[1]/fare_by_survived[0]):.1f} lần người tử vong ({fare_by_survived[0]:.2f}).")
```

## Câu hỏi 4: Gia đình đông người có ảnh hưởng đến khả năng sống sót không?

```python
# TODO

survival_by_family = df.groupby('family_size')['survived'].mean()
display(survival_by_family.round(4))
print(f"\n=> Gia đình nhỏ (2-4 người) có tỷ lệ sống sót cao nhất ({survival_by_family.loc[2:4].mean()*100:.1f}%), đi đơn lẻ hoặc gia đình quá đông tỷ lệ sống sót thấp hơn.")
```

## Câu hỏi 5: Cảng lên tàu (embark_town) nào có tỷ lệ sống sót cao nhất?

```python
# TODO

survival_by_embark = df.groupby('embark_town')['survived'].mean()
display(survival_by_embark.sort_values(ascending=False).round(4))
print(f"\n=> Cảng {survival_by_embark.idxmax()} có tỷ lệ sống sót cao nhất ({survival_by_embark.max()*100:.1f}%).")
```
